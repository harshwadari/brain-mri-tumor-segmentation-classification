# libraries

```python
# ============================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import joblib
```

# paths

```python
# ============================================================
# CELL 2: DEFINE FILE PATHS
# ============================================================

# ------------------------------------------------------------
# GLCM FEATURES
# ------------------------------------------------------------

GLCM_TRAIN_PATH = Path(
    r"C:\Users\harsh\OneDrive\Documents\BE_Major_Project\DATASET\Features(1)\glcm_train.csv"
)

GLCM_TEST_PATH = Path(
    r"C:\Users\harsh\OneDrive\Documents\BE_Major_Project\DATASET\Features(1)\glcm_test.csv"
)


# ------------------------------------------------------------
# DWT FEATURES
# ------------------------------------------------------------

DWT_TRAIN_PATH = Path(
    r"C:\Users\harsh\OneDrive\Documents\BE_Major_Project\DATASET\Features_DWT\dwt_train.csv"
)

DWT_TEST_PATH = Path(
    r"C:\Users\harsh\OneDrive\Documents\BE_Major_Project\DATASET\Features_DWT\dwt_test.csv"
)


# ------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------

OUTPUT_DIR = Path(
    r"C:\Users\harsh\OneDrive\Documents\BE_Major_Project\DATASET\Features_DWT"
)

# Create directory if it does not exist
OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)


print("Output directory:")
print(OUTPUT_DIR)
```

# Load glcm and dwt

```python
# ============================================================
# CELL 3: LOAD GLCM AND DWT DATASETS
# ============================================================

glcm_train = pd.read_csv(GLCM_TRAIN_PATH)
glcm_test = pd.read_csv(GLCM_TEST_PATH)

dwt_train = pd.read_csv(DWT_TRAIN_PATH)
dwt_test = pd.read_csv(DWT_TEST_PATH)


print("GLCM Train Shape:", glcm_train.shape)
print("GLCM Test Shape :", glcm_test.shape)

print("DWT Train Shape :", dwt_train.shape)
print("DWT Test Shape  :", dwt_test.shape)
```

# inspect columns

```python
# ============================================================
# CELL 4: INSPECT DATASET COLUMNS
# ============================================================

print("\nGLCM columns:")
print(glcm_train.columns.tolist())

print("\nDWT columns:")
print(dwt_train.columns.tolist())
```

# check metadata

```python
# ============================================================
# CELL 5: CHECK METADATA COLUMNS
# ============================================================

print("GLCM metadata:")
print(glcm_train[["filename", "class"]].head())

print("\nDWT metadata:")
print(dwt_train[["filename", "class"]].head())
```

# Verify train/test image matching

```python
# ============================================================
# CELL 6: VERIFY GLCM AND DWT IMAGE MATCHING
# ============================================================

# Check that GLCM and DWT contain the same images

glcm_train_names = set(glcm_train["filename"])
dwt_train_names = set(dwt_train["filename"])

glcm_test_names = set(glcm_test["filename"])
dwt_test_names = set(dwt_test["filename"])


print(
    "Common training images:",
    len(glcm_train_names.intersection(dwt_train_names))
)

print(
    "Common testing images:",
    len(glcm_test_names.intersection(dwt_test_names))
)


# Check differences

print(
    "\nGLCM train images missing in DWT:",
    len(glcm_train_names - dwt_train_names)
)

print(
    "DWT train images missing in GLCM:",
    len(dwt_train_names - glcm_train_names)
)

print(
    "\nGLCM test images missing in DWT:",
    len(glcm_test_names - dwt_test_names)
)

print(
    "DWT test images missing in GLCM:",
    len(dwt_test_names - glcm_test_names)
)
```

# Align GLCM and DWT using image name

```python
# ============================================================
# CELL 7: ALIGN GLCM AND DWT DATASETS
# ============================================================
#
# We must make sure that the GLCM features and DWT features
# belong to the SAME MRI image before fusion.
#
# image_name is used as the unique identifier.
# ============================================================

# Merge training datasets

train_merged = pd.merge(
    glcm_train,
    dwt_train,
    on=["filename", "class"],
    how="inner",
    suffixes=("_glcm", "_dwt")
)


# Merge testing datasets

test_merged = pd.merge(
    glcm_test,
    dwt_test,
    on=["filename", "class"],
    how="inner",
    suffixes=("_glcm", "_dwt")
)


print("Merged Train Shape:", train_merged.shape)
print("Merged Test Shape :", test_merged.shape)
```

# Identify GLCM and DWT feature columns

```python
# ============================================================
# CELL 8: IDENTIFY GLCM AND DWT FEATURE COLUMNS
# ============================================================

# ------------------------------------------------------------
# GLCM FEATURES
# ------------------------------------------------------------

glcm_feature_columns = [
    col
    for col in glcm_train.columns
    if col not in ["filename", "class"]
]


# ------------------------------------------------------------
# DWT FEATURES
# ------------------------------------------------------------

dwt_feature_columns = [
    col
    for col in dwt_train.columns
    if col not in ["filename", "class"]
]


print("Number of GLCM features:", len(glcm_feature_columns))
print("Number of DWT features :", len(dwt_feature_columns))
```

# Extract feature matrices

```python
# ============================================================
# CELL 9: EXTRACT GLCM AND DWT FEATURE MATRICES
# ============================================================

# ------------------------------------------------------------
# TRAINING DATA
# ------------------------------------------------------------

X_glcm_train = train_merged[
    glcm_feature_columns
]

X_dwt_train = train_merged[
    dwt_feature_columns
]


# ------------------------------------------------------------
# TESTING DATA
# ------------------------------------------------------------

X_glcm_test = test_merged[
    glcm_feature_columns
]

X_dwt_test = test_merged[
    dwt_feature_columns
]


# ------------------------------------------------------------
# LABELS
# ------------------------------------------------------------

y_train = train_merged["class"]

y_test = test_merged["class"]


print("GLCM Train:", X_glcm_train.shape)
print("DWT Train :", X_dwt_train.shape)

print("GLCM Test :", X_glcm_test.shape)
print("DWT Test  :", X_dwt_test.shape)
```

# Check for missing values

```python
# ============================================================
# CELL 10: CHECK FOR MISSING VALUES
# ============================================================

print("GLCM Train NaN:", X_glcm_train.isna().sum().sum())
print("DWT Train NaN :", X_dwt_train.isna().sum().sum())

print("GLCM Test NaN :", X_glcm_test.isna().sum().sum())
print("DWT Test NaN  :", X_dwt_test.isna().sum().sum())
```

# Fuse GLCM + DWT

```python
# ============================================================
# CELL 11: FUSE GLCM AND DWT FEATURES
# ============================================================
#
# We combine the GLCM and DWT features horizontally.
#
# Example:
#
# GLCM = [G1, G2, G3, ...]
# DWT  = [D1, D2, D3, ...]
#
# Fused = [G1, G2, G3, ..., D1, D2, D3, ...]
#
# No scaling is done yet.
# ============================================================

# Extract GLCM feature columns
glcm_features = [
    col for col in glcm_train.columns
    if col not in ["filename", "class"]
]

# Extract DWT feature columns
dwt_features = [
    col for col in dwt_train.columns
    if col not in ["filename", "class"]
]


# ------------------------------------------------------------
# Create fused TRAIN dataset
# ------------------------------------------------------------

fusion_train = pd.merge(
    glcm_train[["filename", "class"] + glcm_features],
    dwt_train[["filename", "class"] + dwt_features],
    on=["filename", "class"],
    how="inner"
)


# ------------------------------------------------------------
# Create fused TEST dataset
# ------------------------------------------------------------

fusion_test = pd.merge(
    glcm_test[["filename", "class"] + glcm_features],
    dwt_test[["filename", "class"] + dwt_features],
    on=["filename", "class"],
    how="inner"
)


print("Fusion Train Shape:", fusion_train.shape)
print("Fusion Test Shape :", fusion_test.shape)
```

# verify fusion

```python
# ============================================================
# CELL 12: VERIFY FUSED DATA
# ============================================================

print("Number of GLCM features:", len(glcm_features))
print("Number of DWT features :", len(dwt_features))

print(
    "Expected fused features:",
    len(glcm_features) + len(dwt_features)
)

print(
    "Actual fused features:",
    fusion_train.shape[1] - 2
)


print("\nFirst 5 rows of fused training data:")
display(fusion_train.head())
```

# Save raw fused train/test

```python
# ============================================================
# CELL 13: SAVE FUSED TRAIN AND TEST DATA
# ============================================================
#
# These files contain:
# image_name + class + GLCM features + DWT features
#
# They are saved before PCA so we can inspect/use the complete
# fused representation.
# ============================================================

fusion_train_path = OUTPUT_DIR / "fusion_train.csv"
fusion_test_path = OUTPUT_DIR / "fusion_test.csv"


fusion_train.to_csv(
    fusion_train_path,
    index=False
)

fusion_test.to_csv(
    fusion_test_path,
    index=False
)


print("Fusion train saved to:")
print(fusion_train_path)

print("\nFusion test saved to:")
print(fusion_test_path)
```

# Separate features and labels

```python
# ============================================================
# CELL 14: SEPARATE FEATURES AND LABELS
# ============================================================

# Features only
X_fusion_train = fusion_train.drop(
    columns=["filename", "class"]
)

X_fusion_test = fusion_test.drop(
    columns=["filename", "class"]
)


# Labels
y_train = fusion_train["class"]
y_test = fusion_test["class"]


print("X Fusion Train:", X_fusion_train.shape)
print("X Fusion Test :", X_fusion_test.shape)

print("y Train:", y_train.shape)
print("y Test :", y_test.shape)
```

# Scale the fused features

```python
# ============================================================
# CELL 15: SCALE FUSED FEATURES
# ============================================================
#
# We use ONE StandardScaler.
#
# IMPORTANT:
# The scaler is fitted ONLY on training data.
# Test data is transformed using the same scaler.
# ============================================================

from sklearn.preprocessing import StandardScaler

fusion_scaler = StandardScaler()


# Fit on TRAIN and transform TRAIN
X_fusion_train_scaled = fusion_scaler.fit_transform(
    X_fusion_train
)


# Transform TEST using the TRAIN scaler
X_fusion_test_scaled = fusion_scaler.transform(
    X_fusion_test
)


print("Scaled Fusion Train:", X_fusion_train_scaled.shape)
print("Scaled Fusion Test :", X_fusion_test_scaled.shape)
```

# Save the ONE scaler pickle

```python
# ============================================================
# CELL 16: SAVE FUSION SCALER
# ============================================================

import joblib

fusion_scaler_path = OUTPUT_DIR / "fusion_scaler.pkl"


joblib.dump(
    fusion_scaler,
    fusion_scaler_path
)


print("Fusion scaler saved:")
print(fusion_scaler_path)
```

# Create scaled fusion DataFrames

```python
# ============================================================
# CELL 17: CREATE SCALED FUSION DATAFRAMES
# ============================================================

fusion_feature_columns = [
    f"F{i + 1}"
    for i in range(X_fusion_train_scaled.shape[1])
]


# ------------------------------------------------------------
# TRAIN
# ------------------------------------------------------------

fusion_train_scaled_df = pd.DataFrame(
    X_fusion_train_scaled,
    columns=fusion_feature_columns
)

fusion_train_scaled_df.insert(
    0,
    "filename",
    fusion_train["filename"].values
)

fusion_train_scaled_df.insert(
    1,
    "class",
    fusion_train["class"].values
)


# ------------------------------------------------------------
# TEST
# ------------------------------------------------------------

fusion_test_scaled_df = pd.DataFrame(
    X_fusion_test_scaled,
    columns=fusion_feature_columns
)

fusion_test_scaled_df.insert(
    0,
    "filename",
    fusion_test["filename"].values
)

fusion_test_scaled_df.insert(
    1,
    "class",
    fusion_test["class"].values
)


print(fusion_train_scaled_df.head())
```

# Save final fusion train/test

```python
# ============================================================
# CELL 18: SAVE FINAL SCALED FUSION DATA
# ============================================================

fusion_train_scaled_df.to_csv(
    OUTPUT_DIR / "fusion_train.csv",
    index=False
)

fusion_test_scaled_df.to_csv(
    OUTPUT_DIR / "fusion_test.csv",
    index=False
)


print("Saved:")
print(OUTPUT_DIR / "fusion_train.csv")
print(OUTPUT_DIR / "fusion_test.csv")
```

# Apply pca|

```python
# ============================================================
# CELL 19: APPLY PCA TO FUSED FEATURES
# ============================================================
#
# PCA retains 95% of the variance.
#
# PCA is fitted ONLY on the training data.
# ============================================================

from sklearn.decomposition import PCA


pca_fusion = PCA(
    n_components=100
)


# Fit PCA on training data
X_pca_train = pca_fusion.fit_transform(
    X_fusion_train_scaled
)


# Transform test data using the same PCA
X_pca_test = pca_fusion.transform(
    X_fusion_test_scaled
)


print("Original fused features:", X_fusion_train.shape[1])
print("PCA features:", X_pca_train.shape[1])

print(
    "Variance retained:",
    f"{pca_fusion.explained_variance_ratio_.sum() * 100:.2f}%"
)
```

# Save PCA pickle

```python
# ============================================================
# CELL 20: SAVE PCA MODEL
# ============================================================

pca_path = OUTPUT_DIR / "pca_fusion.pkl"


joblib.dump(
    pca_fusion,
    pca_path
)


print("PCA model saved:")
print(pca_path)
```

# Create PCA train/test

```python
# ============================================================
# CELL 21: CREATE PCA TRAIN AND TEST DATAFRAMES
# ============================================================

pca_columns = [
    f"PC{i + 1}"
    for i in range(X_pca_train.shape[1])
]


# ------------------------------------------------------------
# PCA TRAIN
# ------------------------------------------------------------

pca_fusion_train = pd.DataFrame(
    X_pca_train,
    columns=pca_columns
)

pca_fusion_train.insert(
    0,
    "image_name",
    fusion_train["filename"].values
)

pca_fusion_train.insert(
    1,
    "class",
    fusion_train["class"].values
)


# ------------------------------------------------------------
# PCA TEST
# ------------------------------------------------------------

pca_fusion_test = pd.DataFrame(
    X_pca_test,
    columns=pca_columns
)

pca_fusion_test.insert(
    0,
    "image_name",
    fusion_test["filename"].values
)

pca_fusion_test.insert(
    1,
    "class",
    fusion_test["class"].values
)


print("PCA Train Shape:", pca_fusion_train.shape)
print("PCA Test Shape :", pca_fusion_test.shape)
```

# Save PCA train/test

```python
# # ============================================================
# CELL 22: SAVE PCA FUSION TRAIN AND TEST
# ============================================================

pca_fusion_train.to_csv(
    OUTPUT_DIR / "pca_fusion_train.csv",
    index=False
)

pca_fusion_test.to_csv(
    OUTPUT_DIR / "pca_fusion_test.csv",
    index=False
)


print("PCA Fusion Train saved.")
print("PCA Fusion Test saved.")
```

```python
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

components_list = [10, 20, 30, 40, 50, 60, 80, 100]

for n in components_list:

    pca = PCA(n_components=n)

    X_train_pca = pca.fit_transform(X_fusion_train_scaled)
    X_test_pca = pca.transform(X_fusion_test_scaled)

    clf = SVC(kernel="rbf")

    clf.fit(X_train_pca, y_train)

    y_pred = clf.predict(X_test_pca)

    accuracy = accuracy_score(y_test, y_pred)

    print(
        f"Components: {n:3d} | "
        f"Accuracy: {accuracy * 100:.2f}% | "
        f"Variance: {pca.explained_variance_ratio_.sum() * 100:.2f}%"
    )
```
